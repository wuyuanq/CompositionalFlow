```python
n_observables=17
l = 8
d = 2
n_d = 2**l+1
(2**l+1)**d
print l,d,n_d
filenames = ['P.txt','z1.txt','z2.txt','xW1.txt','xW2.txt','xN1.txt','xN2.txt',\
             'xiW.txt','xiN.txt','densiW.txt','densiN.txt','sW.txt','v1.txt',\
             'v2.txt','Cf.txt','liquid.txt','gas.txt']
```

```python
def read_raw_file(filename,n_observables):
    f= open(filename,'r')
    f_out = open(filename+'.no_bool','w')
    
    for line in f:
        if line.find('F')>=0:
            f_out.write('0.0\n')
        elif line.find('T')>=0:
            f_out.write('1.0\n')
        elif (line.find('+')>0) and (line.find('E')<0):

            buf=list(line)
            buf[line.find('+')-1]='E'         
            f_out.write("".join(buf))
        elif (line.find('-')>0) and (line.find('E')<0):

            buf=list(line)
            buf[line.find('-')-1]='E'         
            f_out.write("".join(buf))
        else:
            f_out.write(line)
        
    f.close()
    f_out.close()
    
    raw_data = loadtxt(filename+'.no_bool')
    assert(n_d*n_d*n_observables==raw_data.size)
    raw_data_transformed = raw_data.reshape((n_d,n_d,n_observables))
    return raw_data_transformed 
```

```python
ref_data= read_raw_file('../raw_data/data_from_old_laptop/fullgrid_comparison/fullgrid_257.txt',17)
print ref_data.shape
```

```python
full_data = []
for fname in filenames:
    buf = loadtxt('data_for_kaust/fullgrid_output/2phase/'+fname).reshape((129,129))
    full_data.append(buf)

full_data=array(full_data).transpose((1,2,0))
print full_data.shape
```

```python
def interpolate(full_data):
    new_shape=array(full_data.shape)
    new_shape[:-1]=tuple((new_shape[:-1]-1)*2+1)
    full_data_large=zeros(new_shape)
    # full_data_large=zeros(ref_data.shape)
    full_data_large[::2,::2,:] = full_data
    full_data_large[1::2,::2,:] = 0.5*(full_data[:-1,:,:]+full_data[1:,:,:])
    full_data_large[::2,1::2,:] = 0.5*(full_data[:,:-1,:]+full_data[:,1:,:])
    full_data_large[1::2,1::2,:] = 0.25*(full_data[:-1,:-1,:]+full_data[:-1,1:,:]+full_data[1:,:-1,:]+full_data[1:,1:,:])
    return full_data_large

full_data_large=interpolate(full_data)
```

```python
mask =( full_data_large[:,:,-2]>0.0) * (full_data_large[:,:,-2]<1.0)

mask = logical_or(mask,( full_data_large[:,:,-1]>0.0) * (full_data_large[:,:,-1]<1.0))
mask[:,:2]=True
mask[:2,:]=True
mask[-2:,:]=True
mask[:,-2:]=True
print mask.shape
# mask= logical_not(mask)
pcolormesh(mask)
```

```python
obs=9
diff = ref_data[:,:,obs]-full_data_large[:,:,obs]
diff[mask]=0.0

## detect discont
pcolormesh(diff)
colorbar()
```

```python
L1_rel_ref_full=[]

for i in range(3,15):
    diff = abs((ref_data[:,:,i]-full_data_large[:,:,i])/ref_data[:,:,i])
    diff[mask]=0.0
    diff[full_data_large[:,:,i]==0.0]=0.0
    sum_diff = diff.sum()
    L1_rel_ref_full.append(sum_diff/diff.size)

    print '{:2}: relative L1 error norm: {: 3.10e}'.format(i,L1_rel_ref_full[-1])
    
```
