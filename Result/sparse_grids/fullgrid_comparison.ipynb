```python
n_observables=17
l = 8
d = 2
n_d = 2**l+1
(2**l+1)**d
print l,d,n_d
filenames = ['P.txt','z1.txt','z2.txt','xW1.txt','xW2.txt','xN1.txt','xN2.txt',\
             'xiW.txt','xiN.txt','densiW.txt','densiN.txt','sW.txt','v1.txt',\
             'v2.txt','Cf.txt','liquid.txt','gas.txt']
```

```python
def read_raw_file(filename,n_observables):
    f= open(filename,'r')
    f_out = open(filename+'.no_bool','w')
    
    for line in f:
        if line.find('F')>=0:
            f_out.write('0.0\n')
        elif line.find('T')>=0:
            f_out.write('1.0\n')
        elif (line.find('+')>0) and (line.find('E')<0):

            buf=list(line)
            buf[line.find('+')-1]='E'         
            f_out.write("".join(buf))
        elif (line.find('-')>0) and (line.find('E')<0):

            buf=list(line)
            buf[line.find('-')-1]='E'         
            f_out.write("".join(buf))
        else:
            f_out.write(line)
        
    f.close()
    f_out.close()
    
    raw_data = loadtxt(filename+'.no_bool')
    assert(n_d*n_d*n_observables==raw_data.size)
    raw_data_transformed = raw_data.reshape((n_d,n_d,n_observables))
    return raw_data_transformed 
```

```python
real_data = read_raw_file('./data_for_kaust/fullgrid_comparison/fullgrid_257.txt',n_observables)
dummy_data = read_raw_file('./data_for_kaust/fullgrid_comparison/dummyfullgrid_257.txt',n_observables)
```
