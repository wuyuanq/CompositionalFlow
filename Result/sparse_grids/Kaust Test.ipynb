```python
def read_raw_file(filename,n_observables):
    f= open(filename,'r')
    f_out = open(filename+'.no_bool','w')
    
    for line in f:
        if line.find('F')>=0:
            f_out.write('0.0\n')
        elif line.find('T')>=0:
            f_out.write('1.0\n')
        else:
            f_out.write(line)
        
    f.close()
    f_out.close()
    
    raw_data = loadtxt(filename+'.no_bool')
    assert(n_d*n_d*n_observables==raw_data.size)
    raw_data_transformed = raw_data.reshape((n_d,n_d,n_observables))
    return raw_data_transformed    
```

```python
import sys

##path on ckow laptop
#sys.path.append('/home/ckow/work/sgpp_coding_days/workspace/sgpp/bin/')

##path on ckow office computer
sys.path.append('/work_fast/kowitz/sgpp/bin')
sys.path.append('/home/ckow/work/sgpp/bin')
import copy

from pysgpp import DataVector, Grid, createOperationHierarchisation, createOperationEval,SurplusRefinementFunctor,GridGenerator
```

```python
sum_points=0
erglist=[]
pointlist = []
for n_observable in range(2,17):

    # create a two-dimensional piecewise bi-linear grid
    dim = 2
    grid = Grid.createLinearBoundaryGrid(dim)
    #grid = Grid.createLinearGrid(dim)
    
    gridStorage = grid.getStorage()
    #print "dimensionality:         %d" % (gridStorage.dim())
    
    # create regular grid
    level = 2
    gridGen = grid.createGridGenerator()
    gridGen.regular(level)
    #print "number of grid points:  %d" % (gridStorage.size())
    
    # create coefficient vector
    alpha = DataVector(gridStorage.size())
    alpha.setAll(0.0)
   # print "length of alpha-vector: %d" % (len(alpha))
    #print l
    
    # set function values in alpha
    #f = lambda x0, x1: 16.0 * (x0-1)*x0 * (x1-1)*x1
    for i in xrange(gridStorage.size()):
        gp = gridStorage.get(i)
        #alpha[i] = f(gp.abs(0), gp.abs(1))
        alpha[i] = raw_data_transformed[int(gp.abs(0)*(n_d-1)),int(gp.abs(1)*(n_d-1)),n_observable]
 #   print alpha
    
    # hierarchize
    createOperationHierarchisation(grid).doHierarchisation(alpha)
  #  print alpha

    print grid.getSize()
    
    for it in range(20):
        #gridGen.refineMaxLevel(SurplusRefinementFunctor(alpha,gridGen.getNumberOfRefinablePointsToMaxLevel(l),0.),l)
        #gridGen.refine(SurplusRefinementFunctor(alpha,gridGen.getNumberOfRefinablePoints()/10,0.))
        
        
        gridGen.refineMaxLevel(SurplusRefinementFunctor(alpha,gridGen.getNumberOfRefinablePointsToMaxLevel(l)/5,0.),l) ##used for producing the KAUST output
       
        #gridGen.refine(SurplusRefinementFunctor(alpha,1,0.))
        
        #print 'refinable points',gridGen.getNumberOfRefinablePoints()
        #print gridStorage.size()
        
        alpha.resize(gridStorage.size())
        for i in xrange(gridStorage.size()):
            gp = gridStorage.get(i)
            #alpha[i] = f(gp.abs(0), gp.abs(1))
            alpha[i] = raw_data_transformed[int(gp.abs(0)*(n_d-1)),int(gp.abs(1)*(n_d-1)),n_observable]
            assert(gp.abs(0)*(n_d-1)-int(gp.abs(0)*(n_d-1))<0.001)
        
        # hierarchize
        createOperationHierarchisation(grid).doHierarchisation(alpha)
        #print alpha.array()
    
    print 'points after refinement',gridStorage.size()
    sum_points+=gridStorage.size()
    
    n_points = gridStorage.size()
    points = zeros((n_points,2))
    for i in range(n_points):
        gp = gridStorage.get(i)
        points[i,:]=(gp.abs(0),gp.abs(1))
        
    pointlist.append(points)
    

    
    opEval = createOperationEval(grid)
    
    X,Y = mgrid[0:1:n_d*1j,0:1:n_d*1j]
    points = zip(X.flatten(),Y.flatten())
    erg = zeros((n_d,n_d))
    for p in points:
        erg[int(p[0]*(n_d-1)),int(p[1]*(n_d-1))] = opEval.eval(alpha,DataVector(p))
        
    savetxt('data_for_kaust/sparse_grid_output/'+filenames[n_observable],erg,delimiter= '\n',newline='\n')
    erglist.append(erg)
```
