```python
def read_binary_file(filename,n_components):
    f= open(filename,'rb')
    data=[]
    count=0
    
    while(1):

        buf_data=zeros(n_components)
        for i in range(n_components-2):
            buf = f.read(8)
            try:
                n = struct.unpack('d',buf)
            except:
                f.close()
                res=int(round(roots((1,1,0,-2*len(data)))[2].real))
                print 'found eof, res: ',res
                return array(data).reshape((9,-1,n_components))
            
            
            buf_data[i]=n[0]
        for i in range(n_components-2,n_components):
            buf = f.read(4)
            n = struct.unpack('i',buf)
            buf_data[i]=n[0]

        data.append(buf_data)
    f.close()
    return None
    
    
```

```python
##for p9 only
def embed(raw_data):
    data = zeros((9,257,257,raw_data.shape[2]))
    for i in range(9):
        for j in range(raw_data.shape[1]):
            z1=raw_data[i,j,1]
            z2=raw_data[i,j,2]
            
            z1_ind=int(round(z1*(256)))
            z2_ind=int(round(z2*(256)))

            data[i,z1_ind,z2_ind,:]=raw_data[i,j,:]
    return data
```

```python
all_data=embed(data)
```
