```python
def clean_history(filename,file_out):
    traj_raw = loadtxt(filename)
    bin_outfile = open(file_out,'wb')
    traj_raw.tofile(bin_outfile)
    bin_outfile.close()
```

```python
root_folder = './clean_data/'
case_folders = ['liquidcase/','gascase/','2phasecase1/','2phasecase2/']
file_types = ['-true','-full','-sparse']
full_grid_folders=['fullgrid_output','sparsegrid_output']
hist_files = ['historycurve11','historycurve140','historycurve2020']
recovery_file = ['recovery']

#true_files = [folder+'trueflashhistory/historycurve11.txt',folder+'trueflashhistory/historycurve140.txt',folder+'trueflashhistory/historycurve2020.txt']
#dummy_fullgrid_files = [folder+'dummyhistory/historycurve11-full.txt',folder+'dummyhistory/historycurve140-full.txt',folder+'dummyhistory/historycurve2020-full.txt']
#dummy_sparse_files = [folder+'dummyhistory/historycurve11-sparse.txt',folder+'dummyhistory/historycurve140-sparse.txt',folder+'dummyhistory/historycurve2020-sparse.txt']
#recovery_files = [folder+'trueflashhistory/recovery.txt',folder+'dummyhistory/recovery-full.txt',folder+'dummyhistory/recovery-sparse.txt']
```

```python
def read_file(f):
    infile = open(f,'rb')
    buf=fromfile(infile).reshape((-1,3))
    infile.close()
    return buf
```
