```python
f=open('data_for_kaust/fullgrid.txt','r')
f_out=open('data_for_kaust/fullgrid_transformed.csv','w')

n_obs=21

i=0
for line in f:
    if i==0:
        f_out.write(str((float(line.strip())-1984000.)/128000.)+';') 
    elif line.find('F')>=0:
        f_out.write('0.00;')
    elif line.find('T')>=0:
        f_out.write('1.00;')
    else:
        f_out.write(line.strip()+';')
    i+=1
    if i==n_obs:
        i=0
        f_out.write('\n')
        
f.close()
f_out.close()
        
    

```
